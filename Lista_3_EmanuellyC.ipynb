**1. Guardando e analisando notas**

LISTA — exemplo resolvido


**RACIOCÍNIO / ORIENTAÇÕES**
1. Crie uma lista vazia para as notas.
2. Repita 5 leituras e use append() para armazenar cada nota.
3. Percorra a lista para calcular a soma e a média.
4. Percorra novamente para contar quantas notas são >= média.
5. Mostre a lista, a média e a quantidade.

```python
notas = []
for i in range(5):
  nota = float(input("Digite a nota:"))
  notas.append(nota)
  soma = 0
  for nota in notas:
    soma += nota
    media = soma / len(notas)
    acima_ou_igual = 0
    for nota in notas:
      if nota >= media:
        acima_ou_igual += 1
print("Notas:", notas)
print("Média:", media)
print("Notas >= média", acima_ou_igual)
```

**4. Lista de números: pares e ímpares**

LISTA

**RACIOCÍNIO / ORIENTAÇÕES**

1.   Crie uma lista vazia para todos os números, para os números pares e números ímpares.
2.   Repita 10 leituras e use append() para armazenar cada número.
3.   Percorra a lista original utilizando o módulo %.
4.   Mostre as três listas.

```python
numeros = []
pares = []
impares = []
for i in range(10):
  numero = int(input("Digite o número:"))
  numeros.append(numero)
for numero in numeros:
    if numero % 2 == 0:
       pares.append(numero)
    else:
        impares.append(numero)
print("Números", numeros)
print("Números pares", pares)
print("Números ímpares", impares)
```

LISTA

**RACIOCÍNIO**

3.   Aramazene SOMENTE nomes que ainda não foram digitados.
4.   Exiba a lista e a quantidade de nomes diferentes.


```python
nomes = []
while True:
  nome = input("Digite um nome:")
  if nome == "fim" or nome == "Fim" or nome == "FIM":
    break
if nome not in nomes:
  nomes.append(nome)
else:
    print("O número já está cadastrado.")
print("Nomes", nomes)
print("Quantidade de nomes diferentes:", len(nomes))
```

**7. Dias da semana em tupla**

TUPLA

**RACIOCÍNIO**


1.   Crie uma tupla que contenha os 7 dias da semana.
2.   Leia um número de 1 a 7.
3.   Mostre o dia que corresponde aos números de 1 até 7 (caso o número digitado esteja fora desse intervalo, informe que a opção é inválida).


```python
dias_da_semana = (
    "Domingo",
    "Segunda",
    "Terça",
    "Quarta",
    "Quinta",
    "Sexta",
    "Sábado"
)
dias = input("Digite um número de 1 até 7:")
if dias:
  numero = int(dias)
  if 1 <= numero <= 7:
    dia_equivalente = dias_da_semana[numero - 1]
    print("O dia equivalente é:", dia_equivalente)
  else:
      print("Opção inválida. O número deve estar entre 1 e 7.")
else:
    print("Opção inválida. O número deve ser inteiro.")
```

**13. Cadastro de vários alunos**

LISTA + DICIONÁRIO

**RACIOCÍNIO**
1. Crie um dicionário vazio para armazenar os dados.
2. Cadastre 5 alunos.
3. Para cada aluno, leia nome e duas notas, calcule a média  e armazene nome, notas e média em um dicionário.
4. Adicionar uma lista a cada dicionário.
5. Mostre o resultado com nome e média de todos os alunos.

```python
cadastro_alunos = []
for i in range(5):
  print("Cadastro de um aluno:")
  nome = input("Digite o nome do aluno:")
  nota1 = float(input("Digite a nota 1 do aluno:"))
  nota2 = float(input("Digite a nota 2 do aluno:"))
  media = (nota1 + nota2) / 2
  primeiro_aluno = {
      "nome": nome,
      "media": media
  }
  cadastro_alunos.append(primeiro_aluno)
print("Resultados")

for aluno in cadastro_alunos:
    print("Aluno:", aluno['nome'], "Média:", aluno['media'])
```
