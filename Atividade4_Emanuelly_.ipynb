**4. Maior sequência de crescimento**

LISTA


***RACIOCÍNIO***
1. Crie uma lista vazia e leia 10 números inteiros.
2. Crie duas variáveis.
3. Compare o número atual com o que veio antes dele.
4. Se o número atual for maior que o anterior, aumente a contagem de agora em 1.
5. Se não for maior, veja se a contagem de agora é maior que a recorde.
6. Mostre qual foi a maior sequência.

```python
numeros = []
for i in range(10):
    num = int(input("Digite o {i+1}º número: "))
    numeros.append(num)
maior_seq = 1
atual = 1
for i in range(1, len(numeros)):
    if numeros[i] > numeros[i-1]:
        atual += 1
    else:
        atual = 1
    if atual > maior_seq:
        maior_seq = atual
print("A maior sequência estritamente crescente é: ", {maior_seq})
```

**5. Pontos no plano**

LISTA + TUPLA

***RACIOCÍNIO***

1. Ler as variáveis `x` e `y` para descobrir o mais distante utilizando `ponto_mais_distante`
2. Analisar a lista de pontos preenchida
3. Informar eixos e quadrantes
4. Calcular o valor de x² + y²
5. Comparar para achar o maior valor
6. Mostrar os resultados

```python
pontos = []
for i in range(6):
    x = int(input(f"Ponto {i+1} - X: "))
    y = int(input(f"Ponto {i+1} - Y: "))
    pontos.append((x, y))
eixos = q1 = q2 = q3 = q4 = 0
maior_distancia = -1
mais_distante = None
for x, y in pontos:
    if x == 0 or y == 0:
        eixos += 1
    elif x > 0 and y > 0:
        q1 += 1
    elif x < 0 and y > 0:
        q2 += 1
    elif x < 0 and y < 0:
        q3 += 1
    else:
        q4 += 1
    distancia = (x * x) + (y * y)
    if distancia > maior_distancia:
        maior_distancia = distancia
        mais_distante = (x, y)
print("Quadrantes: Q1:", {q1}, "| Q2:", {q2}, "| Q3:", {q3}, "| Q4:", {q4}, "| Eixos:", {eixos})
print("Ponto mais distante da origem: ", {mais_distante})
```

**6. Frequência e palavra mais repetidas**

DICIONÁRIO

**RACIOCÍNIO**


1. Crie um dicionário.
2. Leia 10 palavras.
3. Mostre as quantidades e todas as palavras empatadas na maior frequência.



```python
palavras_lidas = []
print("Digite 10 palavras:")
for i in range(10):
    palavra = input("Palavra",{i+1}, ": ")
    palavras_lidas.append(palavra)
frequencias = {}
for pal in palavras_lidas:
    if pal in frequencias:
        frequencias[pal] += 1
    else:
        frequencias[pal] = 1
print("Dicionário de frequências:", frequencias)
maior_freq = -1
for pal in frequencias:
    if frequencias[pal] > maior_freq:
        maior_freq = frequencias[pal]
empatadas = []
for pal in frequencias:
    if frequencias[pal] == maior_freq:
        empatadas.append(pal)

print("Maior frequência encontrada: ", {maior_freq})
print("Palavra(s) empatada(s) na maior frequência: ", {empatadas})
```

**7. Boletim de uma turma**

LISTA + DICIONÁRIO

**RACIOCÍNIO**

1. Cadastre 6 alunos. Para cada um, leia nome e três notas.
2. Calcule a média individual de cada aluno.
3. Calcule a média geral após todos os cadastros.
4. Informe os alunos acima ou iguais à média e busque a maior média.
5. Por fim, exiba os resultados.

```python
turma = []
soma_medias_turma = 0
print("Cadastro do boletim da turma (6 alunos):")
for i in range(6):
    print("Dados do Aluno", {i+1}, ":")
    nome = input("Nome do aluno: ")
    notas = []
    for j in range(3):
        nota = float(input("Digite a nota", {j+1}, ":"))
        notas.append(nota)
    media_aluno = sum(notas) / len(notas)
    aluno_dic = {
        "nome": nome,
        "media": media_aluno
    }
    turma.append(aluno_dic)
    soma_medias_turma += media_aluno
media_geral_turma = soma_medias_turma / len(turma)
alunos_na_media_ou_acima = 0
maior_media = -1
for aluno in turma:
    if aluno["media"] >= media_geral_turma:
        alunos_na_media_ou_acima += 1

    if aluno["media"] > maior_media:
        maior_media = aluno["media"]
alunos_maior_media = []
for aluno in turma:
    if aluno["media"] == maior_media:
        alunos_maior_media.append(aluno["nome"])
print("RESULTADOS DA TURMA")
for aluno in turma:
    print("Nome:", {aluno['nome']}, "| Média:", "{aluno['media']:}")

print("Média Geral da Turma:", {media_geral_turma:})
print("Quantidade de alunos com média maior ou igual à geral da turma:", {alunos_na_media_ou_acima})
print("Aluno(s) com a maior média", ({maior_media}), {', '.join(alunos_maior_media)})
```

**8.Controle de estoque com movimentações**

LISTA + DICIONÁRIO

**RACIOCÍNIO**
1. Crie uma lista e cadastre 5 com código, nome, preço e quantidade.
2. Realize 5 operações (E para Entrada ou S para Saída) e a quantidade.
3. Calcule o valor total de cada produto (multiplicando o preço pela quantidade atual) e mostre o estoque atualizado com o valor total de tudo somado.

```python
estoque = []
for i in range(5):
    dados = input(f"Prod {i+1} (Código Nome Preço Qtd): ").split()
    estoque.append({"cod": int(dados[0]), "nome": dados[1], "preco": float(dados[2]), "qtd": int(dados[3])})

for i in range(5):
    cod, tipo, qtd = input(f"Mov {i+1} (Cod Tipo Qtd): ").split()
    qtd = int(qtd)

    if qtd > 0:
        for p in estoque:
            if p["cod"] == int(cod):
                if tipo.upper() == "E": p["qtd"] += qtd
                elif tipo.upper() == "S" and p["qtd"] >= qtd: p["qtd"] -= qtd

# Exibe resultados e calcula total financeiro
total_geral = sum(p["preco"] * p["qtd"] for p in estoque)
for p in estoque:
    print(f"{p['nome']} - Qtd: {p['qtd']} | Total: R${p['preco']*p['qtd']:.2f}")
print(f"Total Geral: R${total_geral:.2f}")
```

**9. Ranking de jogadores**

LISTA + DICIONÁRIO

**RACIOCÍNIO**
1. Crie uma lista que cadastre 5 jogadores.
2. Armazene as três pontuações de cada jogador.
3. Calcule os resultados (total, média e regularidade) após o armazenamento.
4. Descubra ac maior e menor pontuação por comparação direta.

```python
jogadores = []
for i in range(5):
    nome = input(f"Jogador {i+1}: ")
    pontos = [int(p) for p in input("Digite as 3 pontuações: ").split()]
    jogadores.append({"nome": nome, "pontos": pontos})

maior_tot, menor_dif = -1, float('inf')

# Processa após o armazenamento
for j in jogadores:
    pts = j["pontos"]

    # 3 e 4. Descobre maior e menor nota por comparação (sem max/min)
    maior_p, menor_p = pts[0], pts[0]
    for p in pts:
        if p > maior_p: maior_p = p
        if p < menor_p: menor_p = p

    j["total"] = sum(pts)
    j["media"] = j["total"] / 3
    j["dif"] = maior_p - menor_p

    if j["total"] > maior_tot: maior_tot = j["total"]
    if j["dif"] < menor_dif: menor_dif = j["dif"]

# 5. Trata empates
melhores = [j["nome"] for j in jogadores if j["total"] == maior_tot]
regulares = [j["nome"] for j in jogadores if j["dif"] == menor_dif]

print("\n--- RESULTADOS ---")
for j in jogadores:
    print(f"{j['nome']} - Total: {j['total']} | Média: {j['media']:.2f}")
print(f"Maior total ({maior_tot}): {melhores}")
print(f"Mais regular(es) - menor dif ({menor_dif}): {regulares}")
```

**10. Agenda de contatos sem duplicidade**

LISTA + DICIONÁRIO

**RACIOCÍNIO**

1. Armazene os contatos.
2. Cada contato deve ser um dicionário.
3. Procure se o nome já existe.
4. Mostre o resultado.

```python
agenda = []
while True:
    nome = input("Nome (ou FIM para encerrar): ").strip()
    if nome.upper() == "FIM": break
    existe = False
    for c in agenda:
        if c["nome"].lower() == nome.lower():
            existe = True; break
    if existe:
        print("Erro: Este nome já está cadastrado!")
    else:
        tel = input("Telefone: ")
        agenda.append({"nome": nome, "telefone": tel})
print("Contatos")
while True:
    busca = input("Buscar nome: ").strip()
    if busca.upper() == "FIM": break

    encontrado = False
    for c in agenda:
        if c["nome"].lower() == busca.lower():
            print(f"Telefone de {c['nome']}: {c['telefone']}")
            encontrado = True; break

    if not encontrado: print("Contato não encontrado.")
```

**11. Pesquisa de votos**

**RACIOCÍNIO**
1. Leia votos para 4 opções.
2. Utilize um dicionário para contar os votos.
3. Ao final, exiba o total de cada opção, o percentual de cada uma e a(s) opção(ões) vencedora(s).

```python
votos = {1: 0, 2: 0, 3: 0, 4: 0}
total_votos = 0
while True:
    voto = int(input("Voto de 1 a 4 (ou 0 para encerrar): "))
    if voto == 0: break
    if voto in votos:
        votos[voto] += 1
        total_votos += 1
    else:
        print("Voto inválido!")
maior_v = -1
for op in votos:
    if votos[op] > maior_v: maior_v = votos[op]
vencedores = [op for op in votos if votos[op] == maior_v]
print("\n--- RESULTADO DA VOTAÇÃO ---")
for op, qtd in votos.items():
    pct = (qtd / total_votos * 100) if total_votos > 0 else 0
    print(f"Opção {op}: {qtd} votos ({pct:.1f}%)")
print(f"Opção(ões) vencedora(s) com {maior_v} votos: {vencedores}")
```

**12. Histórico de movimentações**

LISTA + TUPLA + DICIONÁRIO

**RACIOCÍNIO**
1. Use um dicionário para o estoque e realize 8 movimentações.
2. Use tupla para cada movimentação.
3. Use uma lista para o histórico.
4. Mostre estoque final e histórico.

```python
estoque = {101: 50, 102: 30, 103: 20}
historico = []
print(f"Estoque Inicial: {estoque}\nRealize 8 movimentações:")
for i in range(8):
    cod, tipo, qtd = input(f"Mov {i+1}/8 (Cod Tipo Qtd): ").split()
    cod, qtd, tipo = int(cod), int(qtd), tipo.upper()
    status = "Recusada"
    if cod in estoque and qtd > 0:
        if tipo == "E":
            estoque[cod] += qtd
            status = "Sucesso"
        elif tipo == "S" and estoque[cod] >= qtd:
            estoque[cod] -= qtd
            status = "Sucesso"
    historico.append((cod, tipo, qtd, status))
    print(f"Operação: {status}")
print(f"\nEstoque Final: {estoque}")
print("Histórico de Movimentações:")
for mov in historico:
    print(f"Cód: {mov[0]} | Tipo: {mov[1]} | Qtd: {mov[2]} -> {mov[3]}")
```

**13. Análise de vendas por vendedor**

LISTA + DICIONÁRIO

**RACIOCÍNIO**

1. Cadastre 5 vendedores.
2. Calcule o total de cada vendedor percorrendo seus valores.
3. Informe total vendido por vendedor, total geral, vendedor(es) com maior total e quantos ficaram acima da média.
4. Exiba os resultados.

```python
equipe = []
print("--- Cadastro dos Vendedores ---")
for i in range(5):
    nome = input(f"\nNome do vendedor {i+1}: ")
    vendas = []
    for j in range(4):
        valor = float(input(f"Valor da venda {j+1}: R$ "))
        vendas.append(valor)
    ficha = {"nome": nome, "vendas": vendas}
    equipe.append(ficha)
total_geral = 0
for vendedor in equipe:
    soma = 0
    for valor in vendedor["vendas"]:
        soma += valor
    vendedor["total"] = soma
    total_geral += soma
media_equipe = total_geral / 5
maior_total = -1
for vendedor in equipe:
    if vendedor["total"] > maior_total:
        maior_total = vendedor["total"]
melhores = []
acima_da_media = 0

for vendedor in equipe:
    if vendedor["total"] == maior_total:
        melhores.append(vendedor["nome"])
    if vendedor["total"] > media_equipe:
        acima_da_media += 1
print("\n=== Relatório de Vendas ===")
for vendedor in equipe:
    print(f"{vendedor['nome']}: R$ {vendedor['total']:.2f}")

print(f"\nTotal da equipe: R$ {total_geral:.2f}")
print(f"Média da equipe: R$ {media_equipe:.2f}")
print(f"Melhor(es) vendedor(es): {melhores} com R$ {maior_total:.2f}")
print(f"Vendedores acima da média: {acima_da_media}")
```

**14. Catálogo e pedidos**

LISTA + TUPLA + DICIONÁRIO

**RACIOCÍNIO**
1. Cadastre os 5 produtos.
2. Receber os 6 pedidos antes de processar
3. Processar os pedidos na ordem em que foram feitos.
4. Informe quais pedidos foram atendidos ou recusados.
5. Mostre o valor total dos pedidos atendidos.

```python
produtos = []
for i in range(5):
    print(f"\nProduto {i+1}:")
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    preco = float(input("Preço: R$ "))
    estoque = int(input("Estoque: "))
    ficha = {"codigo": codigo, "nome": nome, "preco": preco, "estoque": estoque}
    produtos.append(ficha)
pedidos = []
for i in range(6):
    print(f"\nPedido {i+1}:")
    cod_pedido = int(input("Código do produto: "))
    qtd_pedida = int(input("Quantidade: "))
    pedidos.append((cod_pedido, qtd_pedida))
total_ganho = 0
for cod_pedido, qtd_pedida in pedidos:
    achou = None
    for p in produtos:
        if p["codigo"] == cod_pedido:
            achou = p
            break
    if achou is not None:
        if achou["estoque"] >= qtd_pedida:
            achou["estoque"] -= qtd_pedida
            valor_pedido = qtd_pedida * achou["preco"]
            total_ganho += valor_pedido
            print(f"Pedido de {achou['nome']} (Qtd: {qtd_pedida}): ATENDIDO")
        else:
            print(f"Pedido de {achou['nome']} (Qtd: {qtd_pedida}): RECUSADO (Falta estoque)")
    else:
        print(f"Pedido do código {cod_pedido}: RECUSADO (Produto não existe)")

print(f"\nFaturamento total: R$ {total_ganho:.2f}")
```
